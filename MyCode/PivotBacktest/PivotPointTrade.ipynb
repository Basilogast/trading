```python
trader = PivotPointTrader("oanda.cfg", "EUR_USD", "1min", units=100000)
```

```python
trader.get_most_recent()
trader.stream_data(trader.instrument, stop=10000)
if trader.position != 0:  # if we have a final open position
    close_order = trader.create_order(trader.instrument, units=-trader.position * trader.units, suppress=True, ret=True)
    trader.report_trade(close_order, "GOING NEUTRAL")
    trader.position = 0
```
